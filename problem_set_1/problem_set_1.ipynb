### Problem 1 [Theory]

**Task:**
Let $x, y \in \mathbb{R}^d$. Show that if $\|x\|_2 = \|y\|_2 = 1$ then $\|x - y\|_2^2 = 2 (1 - \cos \theta)$, where $\cos \theta$ is the cosine similarity. Deduce that on the unit sphere Euclidean and cosine geometry induce the same ranking of neighbours.

---

### Assumptions & Notation
1. **Space & Vectors:** Let $x, y \in \mathbb{R}^d$ be $d$-dimensional real vectors ($d \ge 1$).
2. **Inner Product:** The standard Euclidean inner product is defined as:
   $$\langle x, y \rangle = x^\top y = \sum_{i=1}^d x_i y_i$$
3. **Euclidean Norm:** The standard $\ell_2$ norm is $\|u\|_2 = \sqrt{\langle u, u \rangle} = \sqrt{\sum_{i=1}^d u_i^2}$.
4. **Unit Sphere Condition:** Both vectors lie on the unit hypersphere $\mathbb{S}^{d-1} = \{u \in \mathbb{R}^d : \|u\|_2 = 1\}$, so:
   $$\|x\|_2 = 1 \quad \text{and} \quad \|y\|_2 = 1$$
5. **Cosine Similarity:** For non-zero vectors, the cosine similarity corresponds to the cosine of the angle $\theta \in [0, \pi]$ between them:
   $$\cos \theta(x, y) = \frac{\langle x, y \rangle}{\|x\|_2 \|y\|_2}$$
   Under the unit norm condition $\|x\|_2 = \|y\|_2 = 1$, this simplifies directly to:
   $$\cos \theta(x, y) = \langle x, y \rangle$$

---

### Formal Proposition
1. **Proposition 1.1 (Norm Identity):**
   $$\|x - y\|_2^2 = 2\bigl(1 - \cos \theta(x, y)\bigr)$$
2. **Proposition 1.2 (Equivalence of Neighbour Rankings):**
   Let $q \in \mathbb{S}^{d-1}$ be a fixed query point and let $x_1, x_2 \in \mathbb{S}^{d-1}$ be two arbitrary candidate points on the unit sphere. Then:
   $$\|q - x_1\|_2 \le \|q - x_2\|_2 \iff \cos \theta(q, x_1) \ge \cos \theta(q, x_2)$$
   with equality holding on the left if and only if equality holds on the right. Thus, Euclidean distance and cosine similarity induce identical nearest-neighbour orderings on $\mathbb{S}^{d-1}$.

---

### Step-by-Step Derivation

#### Step 1: Expansion of the squared Euclidean distance
Using the bilinearity and symmetry of the standard Euclidean inner product:
$$\|x - y\|_2^2 = \langle x - y, x - y \rangle = \langle x, x \rangle - 2\langle x, y \rangle + \langle y, y \rangle$$

#### Step 2: Substitution of the unit norm assumption
Since $\|x\|_2^2 = \langle x, x \rangle = 1$ and $\|y\|_2^2 = \langle y, y \rangle = 1$, substituting into the expanded form gives:
$$\|x - y\|_2^2 = 1 - 2\langle x, y \rangle + 1 = 2 - 2\langle x, y \rangle = 2\bigl(1 - \langle x, y \rangle\bigr)$$

#### Step 3: Expression in terms of cosine similarity
Using the definition of cosine similarity on unit vectors, $\langle x, y \rangle = \cos \theta(x, y)$:
$$\|x - y\|_2^2 = 2\bigl(1 - \cos \theta(x, y)\bigr)$$
Taking the non-negative square root yields the exact relationship:
$$\|x - y\|_2 = \sqrt{2\bigl(1 - \cos \theta(x, y)\bigr)}$$

#### Step 4: Strict monotonic transformation and neighbour ranking
Consider the Euclidean distance as a function of cosine similarity $s = \cos \theta \in [-1, 1]$:
$$f(s) = \sqrt{2(1 - s)}$$
Taking the first derivative of $f(s)$ with respect to $s$ for $s \in (-1, 1)$:
$$f'(s) = \frac{\mathrm{d}}{\mathrm{d}s}\left( (2 - 2s)^{1/2} \right) = \frac{1}{2}(2 - 2s)^{-1/2} \cdot (-2) = -\frac{1}{\sqrt{2(1 - s)}} < 0$$
Since $f'(s) < 0$ everywhere on $(-1, 1)$, $f(s)$ is a strictly decreasing continuous function on $[-1, 1]$.

Now, consider a query $q \in \mathbb{S}^{d-1}$ and two candidate neighbours $x_1, x_2 \in \mathbb{S}^{d-1}$ with respective cosine similarities $s_1 = \cos \theta(q, x_1)$ and $s_2 = \cos \theta(q, x_2)$:
- If $s_1 > s_2$, then because $f$ is strictly decreasing:
  $$\|q - x_1\|_2 = f(s_1) < f(s_2) = \|q - x_2\|_2$$
- If $s_1 = s_2$, then:
  $$\|q - x_1\|_2 = f(s_1) = f(s_2) = \|q - x_2\|_2$$
- If $s_1 < s_2$, then:
  $$\|q - x_1\|_2 = f(s_1) > f(s_2) = \|q - x_2\|_2$$

Therefore:
$$\|q - x_1\|_2 \le \|q - x_2\|_2 \iff \cos \theta(q, x_1) \ge \cos \theta(q, x_2)$$
which proves that on the unit sphere, sorting candidates in ascending order of Euclidean distance produces the exact same permutation (ranking) as sorting them in descending order of cosine similarity (or ascending order of cosine distance $1 - \cos \theta$). $\blacksquare$

---

### Sanity Checks & Edge Cases
1. **Identical vectors ($\theta = 0$):**
   $$\cos \theta = 1 \implies \|x - y\|_2^2 = 2(1 - 1) = 0 \implies \|x - y\|_2 = 0$$
   This matches the identity of indiscernibles for Euclidean distance.
2. **Orthogonal vectors ($\theta = \pi/2$):**
   $$\cos \theta = 0 \implies \|x - y\|_2^2 = 2(1 - 0) = 2 \implies \|x - y\|_2 = \sqrt{2}$$
   This matches the Pythagorean theorem in $\mathbb{R}^d$ for two orthogonal unit vectors: $\|x - y\|_2^2 = \|x\|_2^2 + \|y\|_2^2 = 1 + 1 = 2$.
3. **Antipodal vectors ($\theta = \pi$):**
   $$\cos \theta = -1 \implies \|x - y\|_2^2 = 2(1 - (-1)) = 4 \implies \|x - y\|_2 = 2$$
   This is the maximal Euclidean distance on the unit sphere, corresponding to opposite ends of the sphere diameter.
4. **Dimensional consistency ($d=1$):**
   For $d=1$, unit vectors are $x, y \in \{-1, +1\}$.
   - If $x = y = 1$, $\cos \theta = 1$, $\|x - y\|_2^2 = 0 = 2(1 - 1)$.
   - If $x = 1, y = -1$, $\cos \theta = -1$, $\|x - y\|_2^2 = (1 - (-1))^2 = 4 = 2(1 - (-1))$.
   The formula holds across all dimensions $d \ge 1$.


### Problem 2 [Theory]

**Task:**
Define the cosine distance $d_{\cos}(x, y) = 1 - \cos \theta(x, y)$. Give an explicit counterexample in $\mathbb{R}^2$ showing that $d_{\cos}$ violates the triangle inequality, hence is not a metric. Contrast with the angular distance $\arccos(\cos \theta)/\pi$.

---

### Assumptions & Notation
1. **Metric Space Axioms:** A function $d: M \times M \to \mathbb{R}$ on a set $M$ is a metric if and only if for all $u, v, w \in M$:
   - **Non-negativity:** $d(u, v) \ge 0$.
   - **Identity of indiscernibles:** $d(u, v) = 0 \iff u = v$.
   - **Symmetry:** $d(u, v) = d(v, u)$.
   - **Triangle inequality:** $d(u, w) \le d(u, v) + d(v, w)$.
2. **Cosine Similarity:** For any non-zero vectors $u, v \in \mathbb{R}^d \setminus \{\mathbf{0}\}$, the cosine similarity is:
   $$\cos \theta(u, v) = \frac{\langle u, v \rangle}{\|u\|_2 \|v\|_2} \in [-1, 1]$$
   where $\theta(u, v) \in [0, \pi]$ is the angle between $u$ and $v$.
3. **Cosine Distance:** Defined as:
   $$d_{\cos}(u, v) = 1 - \cos \theta(u, v) \in [0, 2]$$
4. **Angular Distance:** Defined as:
   $$d_{\mathrm{ang}}(u, v) = \frac{\theta(u, v)}{\pi} = \frac{\arccos\bigl(\cos \theta(u, v)\bigr)}{\pi} \in [0, 1]$$

---

### Formal Proposition
1. **Proposition 2.1 (Non-metric nature of cosine distance):**
   The function $d_{\cos}$ is **not** a metric on $\mathbb{R}^2 \setminus \{\mathbf{0}\}$ (nor when restricted to the unit circle $\mathbb{S}^1$) because it violates the triangle inequality: there exist vectors $x, y, z \in \mathbb{R}^2$ such that:
   $$d_{\cos}(x, z) > d_{\cos}(x, y) + d_{\cos}(y, z)$$
   *(Note: $d_{\cos}$ also fails the identity of indiscernibles on $\mathbb{R}^d$ because $d_{\cos}(u, cu) = 0$ for any $c > 0$, but even when restricted to the unit sphere $\mathbb{S}^{d-1}$ where identity of indiscernibles holds, the failure of the triangle inequality definitively disqualifies it as a metric).*
2. **Proposition 2.2 (Metric validity of angular distance):**
   The normalized angular distance $d_{\mathrm{ang}}$ is a valid metric on the unit sphere $\mathbb{S}^{d-1}$ (and on rays/directions in $\mathbb{R}^d$), satisfying all metric axioms including the triangle inequality.

---

### Step-by-Step Counterexample & Derivation

#### Step 1: Selection of three vectors in $\mathbb{R}^2$
Choose three unit vectors $x, y, z \in \mathbb{S}^1 \subset \mathbb{R}^2$ corresponding to polar angles $\theta_x = 0$, $\theta_y = \frac{\pi}{4}$ ($45^\circ$), and $\theta_z = \frac{\pi}{2}$ ($90^\circ$):
$$x = \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad y = \begin{bmatrix} \cos(\pi/4) \\ \sin(\pi/4) \end{bmatrix} = \begin{bmatrix} 1/\sqrt{2} \\ 1/\sqrt{2} \end{bmatrix}, \quad z = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$

#### Step 2: Evaluation of pairwise cosine similarities
Using the inner product on unit vectors $\cos \theta(u, v) = u^\top v$:
$$\cos \theta(x, y) = 1 \cdot \frac{1}{\sqrt{2}} + 0 \cdot \frac{1}{\sqrt{2}} = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \approx 0.707107$$
$$\cos \theta(y, z) = \frac{1}{\sqrt{2}} \cdot 0 + \frac{1}{\sqrt{2}} \cdot 1 = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \approx 0.707107$$
$$\cos \theta(x, z) = 1 \cdot 0 + 0 \cdot 1 = 0$$

#### Step 3: Evaluation of pairwise cosine distances
Substitute the similarities into $d_{\cos}(u, v) = 1 - \cos \theta(u, v)$:
$$d_{\cos}(x, y) = 1 - \frac{\sqrt{2}}{2} \approx 0.292893$$
$$d_{\cos}(y, z) = 1 - \frac{\sqrt{2}}{2} \approx 0.292893$$
$$d_{\cos}(x, z) = 1 - 0 = 1.000000$$

#### Step 4: Verification of the triangle inequality violation
Calculate the sum of the two detour distances:
$$d_{\cos}(x, y) + d_{\cos}(y, z) = 2\left(1 - \frac{\sqrt{2}}{2}\right) = 2 - \sqrt{2} \approx 0.585786$$
Now compare $d_{\cos}(x, z)$ with this sum:
$$d_{\cos}(x, z) = 1.000000 > 0.585786 \approx d_{\cos}(x, y) + d_{\cos}(y, z)$$
Since $1 > 2 - \sqrt{2}$, the direct distance between $x$ and $z$ strictly exceeds the path passing through intermediate point $y$:
$$d_{\cos}(x, z) > d_{\cos}(x, y) + d_{\cos}(y, z)$$
Thus, the triangle inequality is explicitly violated, proving that **$d_{\cos}$ is not a metric**. $\blacksquare$

#### Step 5: Geometric reason for the failure
Using the trigonometric half-angle identity:
$$d_{\cos}(\theta) = 1 - \cos \theta = 2\sin^2\left(\frac{\theta}{2}\right)$$
For angles $\theta \in (0, \pi/2)$, the second derivative with respect to $\theta$ is:
$$\frac{\mathrm{d}^2}{\mathrm{d}\theta^2}(1 - \cos \theta) = \cos \theta > 0$$
Because $d_{\cos}(\theta)$ is strictly convex on $[0, \pi/2]$, it is super-additive for angles along the circle:
$$\text{For } \theta_1, \theta_2 > 0 \text{ with } \theta_1 + \theta_2 \le \frac{\pi}{2}: \quad 1 - \cos(\theta_1 + \theta_2) > (1 - \cos \theta_1) + (1 - \cos \theta_2)$$
A distance measure must be sub-additive along straight geodesic paths; strict convexity causes distance to accelerate quadratically with angular separation, breaking the triangle inequality.

---

### Contrast with the Angular Distance

#### 1. Metric Properties of Angular Distance
The angular distance $d_{\mathrm{ang}}(u, v) = \frac{\theta(u, v)}{\pi} = \frac{\arccos(\cos \theta(u, v))}{\pi}$ satisfies all four metric axioms on $\mathbb{S}^{d-1}$:
1. **Non-negativity:** Since the principal range of $\arccos$ is $[0, \pi]$, $d_{\mathrm{ang}}(u, v) \in [0, 1] \ge 0$.
2. **Identity of indiscernibles:** $d_{\mathrm{ang}}(u, v) = 0 \iff \theta(u, v) = 0 \iff u = v$ on $\mathbb{S}^{d-1}$.
3. **Symmetry:** $\cos \theta(u, v) = \cos \theta(v, u) \implies d_{\mathrm{ang}}(u, v) = d_{\mathrm{ang}}(v, u)$.
4. **Triangle inequality:** On the unit hypersphere $\mathbb{S}^{d-1}$ endowed with the standard Riemannian metric, the geodesic distance between two points along the great circle is exactly the angle $\theta(u, v)$. In any Riemannian manifold, geodesic distance satisfies the triangle inequality:
   $$\theta(x, z) \le \theta(x, y) + \theta(y, z)$$
   Dividing across by the constant $\pi > 0$:
   $$\frac{\theta(x, z)}{\pi} \le \frac{\theta(x, y)}{\pi} + \frac{\theta(y, z)}{\pi} \implies d_{\mathrm{ang}}(x, z) \le d_{\mathrm{ang}}(x, y) + d_{\mathrm{ang}}(y, z)$$

#### 2. Evaluation on the Counterexample Vectors
For our specific three vectors $x, y, z$:
$$d_{\mathrm{ang}}(x, y) = \frac{\pi/4}{\pi} = \frac{1}{4} = 0.25$$
$$d_{\mathrm{ang}}(y, z) = \frac{\pi/4}{\pi} = \frac{1}{4} = 0.25$$
$$d_{\mathrm{ang}}(x, z) = \frac{\pi/2}{\pi} = \frac{1}{2} = 0.50$$
Checking the triangle inequality:
$$d_{\mathrm{ang}}(x, z) = 0.50 \le 0.25 + 0.25 = d_{\mathrm{ang}}(x, y) + d_{\mathrm{ang}}(y, z)$$
The inequality holds with exact equality because $y$ lies directly on the minimal geodesic arc between $x$ and $z$.

---

### Sanity Checks & Edge Cases
1. **Collinear vectors in the same direction ($\theta = 0$):**
   $$d_{\cos} = 1 - \cos 0 = 0, \quad d_{\mathrm{ang}} = \frac{0}{\pi} = 0$$
2. **Collinear vectors in opposite directions ($\theta = \pi$):**
   $$d_{\cos} = 1 - \cos \pi = 2, \quad d_{\mathrm{ang}} = \frac{\pi}{\pi} = 1$$
3. **Practical Consequence for Algorithms:**
   Spatial indexing data structures like VP-trees, Ball-trees, and metric-based approximate nearest neighbour algorithms rely on the triangle inequality for branch pruning ($d(q, c) \le d(q, x) + d(x, c)$). Using $d_{\cos}$ causes tree pruning to prune away true nearest neighbours, whereas $d_{\mathrm{ang}}$ (or the Euclidean distance on normalized vectors via Problem 1) preserves correct metric bounding.


### Problem 3 [Theory] (Curse of dimensionality)

**Task:**
For $n$ points drawn i.i.d. uniformly in $[0, 1]^d$, argue that the relative gap between the nearest and farthest point from a fixed query, $(D_{\max}-D_{\min})/D_{\min}$, tends to 0 as $d \to \infty$. Separately, show the fraction of the volume of a $d$-ball within an $\varepsilon$-shell of its surface, $1 - (1 - \varepsilon)^d$, tends to 1. Comment on the consequences for nearest-neighbour search.

---

### Assumptions & Notation
1. **Domain & Query:** Let the ambient space be the unit hypercube $[0, 1]^d$ in dimension $d \ge 1$. Let $q = (q_1, \dots, q_d) \in [0, 1]^d$ be a fixed query point.
2. **Data Distribution:** Consider $n$ points $\{X^{(1)}, \dots, X^{(n)}\}$ drawn independently and identically distributed (i.i.d.) according to the uniform distribution over $[0, 1]^d$. Each coordinate $X^{(i)}_j \sim \mathrm{Uniform}(0, 1)$ independently for $i \in \{1, \dots, n\}$ and $j \in \{1, \dots, d\}$.
3. **Euclidean Distance:** The distance from query $q$ to point $X^{(i)}$ is:
   $$D_i = \|X^{(i)} - q\|_2 = \sqrt{\sum_{j=1}^d (X^{(i)}_j - q_j)^2}$$
4. **Extreme Distances:**
   $$D_{\min} = \min_{1 \le i \le n} D_i \quad \text{and} \quad D_{\max} = \max_{1 \le i \le n} D_i$$
5. **Relative Distance Contrast:**
   $$\mathcal{R}_d = \frac{D_{\max} - D_{\min}}{D_{\min}}$$
6. **Geometry of the $d$-Ball:** Let $B_d(R) = \{x \in \mathbb{R}^d : \|x\|_2 \le R\}$ denote the Euclidean ball of radius $R > 0$ with volume $V_d(R) = C_d R^d$, where $C_d = \frac{\pi^{d/2}}{\Gamma(d/2 + 1)}$. Let $\varepsilon \in (0, 1)$ specify the relative shell thickness.

---

### Formal Proposition
1. **Proposition 3.1 (Distance Concentration / Beyer et al. 1999):**
   For any fixed number of samples $n \ge 2$ and fixed query $q \in [0, 1]^d$:
   $$\frac{D_{\max} - D_{\min}}{D_{\min}} \xrightarrow{p} 0 \quad \text{as } d \to \infty$$
   meaning the relative gap between the farthest and nearest neighbour converges in probability to 0.
2. **Proposition 3.2 (Volume Concentration in Boundary Shell):**
   The fraction of the total volume of $B_d(R)$ contained within the outer boundary shell of thickness $\varepsilon R$ is:
   $$f_d(\varepsilon) = \frac{V_d(R) - V_d\bigl(R(1 - \varepsilon)\bigr)}{V_d(R)} = 1 - (1 - \varepsilon)^d$$
   and satisfies:
   $$\lim_{d \to \infty} f_d(\varepsilon) = 1 \quad \text{for any fixed } \varepsilon \in (0, 1)$$

---

### Step-by-Step Derivation

#### Part A: Relative Gap Between Nearest and Farthest Points

##### Step 1: Distribution of coordinate-wise squared distances
For any fixed point $X^{(i)}$, define the $j$-th squared coordinate distance:
$$Z_{ij} = (X^{(i)}_j - q_j)^2$$
Since $X^{(i)}_j \sim \mathrm{Uniform}(0, 1)$, $Z_{ij}$ are independent random variables bounded in $[0, 1]$.
Compute the mean $\mu_j$ of $Z_{ij}$:
$$\mu_j = \mathbb{E}[Z_{ij}] = \int_0^1 (u - q_j)^2 \, \mathrm{d}u = \left[ \frac{(u - q_j)^3}{3} \right]_0^1 = \frac{(1 - q_j)^3 - (-q_j)^3}{3} = q_j^2 - q_j + \frac{1}{3}$$
Notice that the quadratic $g(q) = q^2 - q + \frac{1}{3}$ achieves its minimum at $q = \frac{1}{2}$ with $g(1/2) = \frac{1}{12}$, and its maximum at $q \in \{0, 1\}$ with $g(0) = g(1) = \frac{1}{3}$.
Thus, for all $j \in \{1, \dots, d\}$:
$$\frac{1}{12} \le \mu_j \le \frac{1}{3}$$
The average coordinate mean is $\bar{\mu}_d = \frac{1}{d} \sum_{j=1}^d \mu_j \in [1/12, 1/3] > 0$.
Similarly, the coordinate variance $\sigma_j^2 = \operatorname{Var}(Z_{ij})$ is bounded: $0 < \sigma_{\min}^2 \le \sigma_j^2 \le 1$. Let $\bar{\sigma}_d^2 = \frac{1}{d}\sum_{j=1}^d \sigma_j^2$.

##### Step 2: Sum of components and asymptotic concentration
The squared Euclidean distance is the sum of $d$ independent bounded random variables:
$$S_{i,d} = D_i^2 = \sum_{j=1}^d Z_{ij}$$
By linearity of expectation and independence:
$$\mathbb{E}[S_{i,d}] = d \bar{\mu}_d, \quad \operatorname{Var}(S_{i,d}) = d \bar{\sigma}_d^2$$
Dividing by $d$:
$$\mathbb{E}\left[\frac{S_{i,d}}{d}\right] = \bar{\mu}_d, \quad \operatorname{Var}\left(\frac{S_{i,d}}{d}\right) = \frac{\bar{\sigma}_d^2}{d} \to 0 \quad \text{as } d \to \infty$$
By Chebyshev's inequality, $\frac{S_{i,d}}{d} - \bar{\mu}_d \xrightarrow{p} 0$, or equivalently:
$$\frac{S_{i,d}}{d \bar{\mu}_d} \xrightarrow{p} 1$$
By the Continuous Mapping Theorem (using $h(t) = \sqrt{t}$ on $t > 0$):
$$\frac{D_i}{\sqrt{d \bar{\mu}_d}} = \sqrt{\frac{S_{i,d}}{d \bar{\mu}_d}} \xrightarrow{p} 1$$

##### Step 3: Extreme value convergence for fixed $n$
For any small tolerance $\delta > 0$, using the union bound across all $n$ data points:
$$\mathbb{P}\left( \max_{1 \le i \le n} \left| \frac{D_i}{\sqrt{d \bar{\mu}_d}} - 1 \right| > \delta \right) \le \sum_{i=1}^n \mathbb{P}\left( \left| \frac{D_i}{\sqrt{d \bar{\mu}_d}} - 1 \right| > \delta \right)$$
By Chebyshev's inequality applied to each term:
$$\mathbb{P}\left( \left| \frac{D_i}{\sqrt{d \bar{\mu}_d}} - 1 \right| > \delta \right) \le \frac{\operatorname{Var}(D_i / \sqrt{d\bar{\mu}_d})}{\delta^2} = \frac{O(1/d)}{\delta^2}$$
Since $n$ is fixed as $d \to \infty$:
$$\lim_{d \to \infty} \mathbb{P}\left( \max_{1 \le i \le n} \left| \frac{D_i}{\sqrt{d \bar{\mu}_d}} - 1 \right| > \delta \right) \le \lim_{d \to \infty} \frac{n \cdot O(1/d)}{\delta^2} = 0$$
Hence, with probability approaching 1:
$$(1 - \delta)\sqrt{d\bar{\mu}_d} \le D_{\min} \le D_{\max} \le (1 + \delta)\sqrt{d\bar{\mu}_d}$$
Subtracting bounds and dividing by $D_{\min}$:
$$\frac{D_{\max} - D_{\min}}{D_{\min}} \le \frac{(1 + \delta)\sqrt{d\bar{\mu}_d} - (1 - \delta)\sqrt{d\bar{\mu}_d}}{(1 - \delta)\sqrt{d\bar{\mu}_d}} = \frac{2\delta}{1 - \delta}$$
Since $\delta > 0$ can be chosen arbitrarily small:
$$\frac{D_{\max} - D_{\min}}{D_{\min}} \xrightarrow{p} 0 \quad \text{as } d \to \infty \quad \blacksquare$$

---

#### Part B: Volume of a $d$-Ball in an $\varepsilon$-Shell

##### Step 1: Volume formula for a $d$-ball
The volume (Lebesgue measure) of a ball of radius $R$ in $\mathbb{R}^d$ is:
$$V_d(R) = \frac{\pi^{d/2}}{\Gamma(d/2 + 1)} R^d = C_d R^d$$

##### Step 2: Volume of the inner concentric ball
Consider the inner ball of radius $R_{\mathrm{in}} = R(1 - \varepsilon)$ where $0 < \varepsilon < 1$:
$$V_d\bigl(R(1 - \varepsilon)\bigr) = C_d \bigl(R(1 - \varepsilon)\bigr)^d = C_d R^d (1 - \varepsilon)^d = (1 - \varepsilon)^d V_d(R)$$

##### Step 3: Volume of the boundary shell and ratio
The $\varepsilon$-shell comprises the annular region between radius $R(1 - \varepsilon)$ and $R$:
$$V_{\mathrm{shell}}(\varepsilon, R) = V_d(R) - V_d\bigl(R(1 - \varepsilon)\bigr) = C_d R^d \bigl( 1 - (1 - \varepsilon)^d \bigr)$$
The fraction of the total ball volume contained in this boundary shell is:
$$f_d(\varepsilon) = \frac{V_{\mathrm{shell}}(\varepsilon, R)}{V_d(R)} = \frac{C_d R^d \bigl( 1 - (1 - \varepsilon)^d \bigr)}{C_d R^d} = 1 - (1 - \varepsilon)^d$$

##### Step 4: Asymptotic limit
Because $0 < \varepsilon < 1$, the base satisfies $0 < 1 - \varepsilon < 1$.
As $d \to \infty$, an exponential with a base strictly less than 1 decays to 0:
$$\lim_{d \to \infty} (1 - \varepsilon)^d = 0$$
Therefore:
$$\lim_{d \to \infty} f_d(\varepsilon) = 1 - 0 = 1 \quad \blacksquare$$

---

### Sanity Checks & Numerical Examples
1. **Numerical verification of shell volume concentration:**
   Let $\varepsilon = 0.05$ (a shell representing merely the outer 5% of radius):
   - For $d = 2$: $1 - (0.95)^2 = 1 - 0.9025 \approx 9.75\%$
   - For $d = 10$: $1 - (0.95)^{10} \approx 40.1\%$
   - For $d = 100$: $1 - (0.95)^{100} \approx 99.41\%$
   - For $d = 1000$: $1 - (0.95)^{1000} \approx 1 - 5.3 \times 10^{-23} \approx 100.00\%$
   Almost all the volume of a high-dimensional ball is packed tightly at its extreme boundary surface.
2. **Dimension $d=1$ check:**
   - Shell fraction: $1 - (1 - \varepsilon)^1 = \varepsilon$, which matches the linear fraction of an interval $[ -R, R ]$.
   - Distance gap for $d=1$: $D_i$ does not concentrate; $D_{\max} - D_{\min}$ is order 1 while $D_{\min}$ is order 1, so the gap does not vanish. Concentration is strictly a high-dimensional phenomenon ($d \to \infty$).

### Consequences for Nearest-Neighbour Search

1. **Loss of Contrast / Meaningless Proximity:**
   Because $\frac{D_{\max} - D_{\min}}{D_{\min}} \to 0$, all pairwise distances converge to nearly the same value ($\approx \sqrt{d\bar{\mu}}$). The discrimination between a "true" nearest neighbour and a random distant point diminishes to noise. Minor measurement errors or uninformative noise dimensions overwhelm meaningful categorical differences.
2. **The "Empty Space" Phenomenon:**
   Because the volume of a ball concentrates entirely in a paper-thin shell at its perimeter, the interior of the high-dimensional space is practically empty. A query point sits in an isolated void, surrounded by candidates situated in an equidistant spherical shell.
3. **Catastrophic Degradation of Spatial Indexing:**
   Geometric data structures designed for fast nearest-neighbour queries (such as $k$-d trees, VP-trees, and Ball-trees) partition space using bounding boxes or bounding hyperspheres. In high dimensions, every bounding sphere overlaps almost entirely with the query search radius. Branch pruning fails, forcing the algorithm to backtrack through virtually every leaf node, causing $O(\log n)$ tree searches to degrade to $O(d \cdot n)$ brute-force linear scans.
4. **Algorithmic Remedy:**
   This phenomenon proves why raw Euclidean distance in high dimensions is flawed for unstructured data (images, text embeddings, audio), directly motivating:
   - **Dimensionality reduction** (PCA, SVD, autoencoders, UMAP) to project data onto low-dimensional intrinsic manifolds where distances retain contrast.
   - **Cosine similarity / angular distances** on normalized embeddings (referencing Problem 1).
   - **Approximate Nearest Neighbour (ANN)** indexing methods (e.g., HNSW, ScaNN, FAISS) that bypass metric bounding hierarchies in favor of graph traversals and vector quantization.


### Problem 4 [Theory] (PCA three ways)

**Task:**
Let $X \in \mathbb{R}^{n \times p}$ be column-centred with SVD $X = U\Sigma V^\top$. Prove that the direction $w$ (with $\|w\| = 1$) maximising the sample variance $\mathrm{Var}(Xw)$ is the top right singular vector $v_1$, and that this coincides with the leading eigenvector of the covariance $\frac{1}{n} X^\top X$. Show further that the rank-$k$ PCA projection minimises the reconstruction error $\sum_{i} \|x_i - \hat{x}_i\|_2^2$.

---

### Assumptions & Notation
1. **Data Matrix:** Let $X \in \mathbb{R}^{n \times p}$ denote the dataset of $n$ observations across $p$ features, where each row $x_i^\top \in \mathbb{R}^{1 \times p}$ is a data point.
2. **Column-Centred Condition:** The feature means are zero:
   $$\frac{1}{n} \mathbf{1}_n^\top X = \mathbf{0}_p^\top \iff \sum_{i=1}^n x_i = \mathbf{0}_p$$
3. **Sample Covariance Matrix:** The $p \times p$ sample covariance matrix of the column-centred data is:
   $$C = \frac{1}{n} X^\top X$$
   which is symmetric and positive semi-definite ($C = C^\top \ge 0$).
4. **Singular Value Decomposition (SVD):**
   $$X = U \Sigma V^\top$$
   where:
   - $U \in \mathbb{R}^{n \times r}$ has orthonormal columns ($U^\top U = I_r$), representing left singular vectors.
   - $V = [v_1, v_2, \dots, v_p] \in \mathbb{R}^{p \times p}$ is an orthogonal matrix ($V^\top V = V V^\top = I_p$), whose columns $v_j$ are the right singular vectors.
   - $\Sigma \in \mathbb{R}^{n \times p}$ is a diagonal matrix of non-negative singular values ordered non-increasingly: $\sigma_1 \ge \sigma_2 \ge \dots \ge \sigma_r \ge 0$, where $r = \min(n, p)$.
5. **Rank-$k$ Projection Matrix:** Let $W_k \in \mathbb{R}^{p \times k}$ have orthonormal columns ($W_k^\top W_k = I_k$). The orthogonal projection of point $x_i$ onto the $k$-dimensional subspace spanned by $W_k$ is $\hat{x}_i = W_k W_k^\top x_i$. In matrix form, $\hat{X} = X W_k W_k^\top$.

---

### Formal Proposition
1. **Proposition 4.1 (Maximum Variance Direction):**
   The unit vector $w^* \in \mathbb{R}^p$ ($\|w^*\|_2 = 1$) that maximizes the sample variance $\operatorname{Var}(Xw)$ is the top right singular vector $v_1$ (up to sign $\pm v_1$):
   $$w^* = \arg\max_{\|w\|_2 = 1} \operatorname{Var}(Xw) = \pm v_1$$
   with maximal variance $\operatorname{Var}(X v_1) = \frac{\sigma_1^2}{n}$.
2. **Proposition 4.2 (Equivalence to Covariance Eigenvector):**
   The top right singular vector $v_1$ is the leading eigenvector of the sample covariance matrix $C = \frac{1}{n} X^\top X$, corresponding to its maximal eigenvalue $\lambda_1 = \frac{\sigma_1^2}{n}$.
3. **Proposition 4.3 (Minimum Reconstruction Error / Best Low-Rank Approximation):**
   The rank-$k$ linear orthogonal projection that minimizes the total squared Euclidean reconstruction error:
   $$\mathcal{E}(W_k) = \sum_{i=1}^n \|x_i - \hat{x}_i\|_2^2 = \|X - X W_k W_k^\top\|_F^2$$
   subject to $W_k^\top W_k = I_k$, is achieved when the columns of $W_k$ span the subspace of the first $k$ right singular vectors $V_k = [v_1, \dots, v_k]$. The minimum reconstruction error is:
   $$\min_{W_k^\top W_k = I_k} \mathcal{E}(W_k) = \sum_{j=k+1}^p \sigma_j^2$$

---

### Step-by-Step Derivation

#### Step 1: Formulation of Sample Variance
Let $w \in \mathbb{R}^p$ be a direction with $\|w\|_2 = 1$. The projected coordinate of sample $x_i$ is $z_i = x_i^\top w$, and the vector of all projected coordinates is $z = Xw \in \mathbb{R}^n$.
Because $X$ is column-centred, the sample mean of $z$ is zero:
$$\bar{z} = \frac{1}{n} \sum_{i=1}^n z_i = \frac{1}{n} \mathbf{1}_n^\top (Xw) = \left(\frac{1}{n}\mathbf{1}_n^\top X\right) w = \mathbf{0}^\top w = 0$$
Therefore, the sample variance of $Xw$ is:
$$\operatorname{Var}(Xw) = \frac{1}{n} \sum_{i=1}^n z_i^2 = \frac{1}{n} \|Xw\|_2^2 = \frac{1}{n} (Xw)^\top (Xw) = w^\top \left(\frac{1}{n} X^\top X\right) w = w^\top C w$$

#### Step 2: Spectral Decomposition of the Covariance via SVD
Substitute the SVD $X = U \Sigma V^\top$ into the covariance matrix $C$:
$$C = \frac{1}{n} X^\top X = \frac{1}{n} (U \Sigma V^\top)^\top (U \Sigma V^\top) = \frac{1}{n} V \Sigma^\top U^\top U \Sigma V^\top$$
Since the columns of $U$ are orthonormal ($U^\top U = I$):
$$C = V \left(\frac{1}{n} \Sigma^\top \Sigma\right) V^\top$$
The matrix $\Sigma^\top \Sigma$ is a $p \times p$ diagonal matrix with diagonal entries $\sigma_1^2, \sigma_2^2, \dots, \sigma_p^2$ (with $\sigma_j = 0$ for $j > \min(n, p)$).
Define the eigenvalues:
$$\lambda_j = \frac{\sigma_j^2}{n} \quad \text{for } j \in \{1, \dots, p\}$$
Since $\sigma_1 \ge \sigma_2 \ge \dots \ge \sigma_p \ge 0$, the eigenvalues are naturally ordered:
$$\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_p \ge 0$$
Thus, the spectral decomposition of $C$ is:
$$C = V \Lambda V^\top = \sum_{j=1}^p \lambda_j v_j v_j^\top$$
Multiplying by any column $v_j$ (using orthonormality $v_l^\top v_j = \delta_{lj}$):
$$C v_j = \left(\sum_{l=1}^p \lambda_l v_l v_l^\top\right) v_j = \lambda_j v_j = \left(\frac{\sigma_j^2}{n}\right) v_j$$
This proves that the right singular vectors $v_j$ of $X$ are precisely the eigenvectors of the covariance matrix $C$, with corresponding eigenvalues $\lambda_j = \frac{\sigma_j^2}{n}$. In particular, the top right singular vector $v_1$ is the leading eigenvector of $C$.

#### Step 3: Maximization of Sample Variance (Rayleigh Quotient)
We seek to solve:
$$\max_{w \in \mathbb{R}^p} w^\top C w \quad \text{subject to } w^\top w = 1$$
Since the eigenvectors $\{v_1, \dots, v_p\}$ form an orthonormal basis of $\mathbb{R}^p$, any unit vector $w$ can be uniquely written as:
$$w = \sum_{j=1}^p \alpha_j v_j = V \alpha, \quad \text{with } \alpha = V^\top w \in \mathbb{R}^p$$
The unit norm constraint becomes:
$$\|w\|_2^2 = w^\top w = \alpha^\top V^\top V \alpha = \alpha^\top I_p \alpha = \sum_{j=1}^p \alpha_j^2 = 1$$
Substitute $w = V\alpha$ into the variance quadratic form:
$$w^\top C w = \alpha^\top V^\top (V \Lambda V^\top) V \alpha = \alpha^\top \Lambda \alpha = \sum_{j=1}^p \lambda_j \alpha_j^2$$
Because $\lambda_1 \ge \lambda_j$ for all $j \in \{1, \dots, p\}$:
$$\sum_{j=1}^p \lambda_j \alpha_j^2 \le \sum_{j=1}^p \lambda_1 \alpha_j^2 = \lambda_1 \sum_{j=1}^p \alpha_j^2 = \lambda_1$$
Equality holds if and only if:
$$\sum_{j=1}^p (\lambda_1 - \lambda_j) \alpha_j^2 = 0 \iff \alpha_j = 0 \quad \text{for all } j \text{ such that } \lambda_j < \lambda_1$$
Assuming $\lambda_1 > \lambda_2$ (a unique leading eigenvalue), this requires $\alpha_1^2 = 1$ and $\alpha_j = 0$ for all $j \ge 2$, which gives:
$$w^* = \pm v_1$$
The maximal variance achieved is:
$$\operatorname{Var}(X w^*) = \lambda_1 = \frac{\sigma_1^2}{n}$$

*(Lagrangian formulation check)*: Define $\mathcal{L}(w, \lambda) = w^\top C w - \lambda(w^\top w - 1)$. Setting the gradient with respect to $w$ to zero:
$$\nabla_w \mathcal{L} = 2 C w - 2\lambda w = \mathbf{0} \implies C w = \lambda w$$
Thus stationary points must be eigenvectors of $C$, with objective value $w^\top C w = \lambda w^\top w = \lambda$, which is globally maximized by selecting the maximal eigenvalue $\lambda_1$ with eigenvector $v_1$.

#### Step 4: Minimization of Reconstruction Error
Let $W_k = [w_1, \dots, w_k] \in \mathbb{R}^{p \times k}$ have orthonormal columns ($W_k^\top W_k = I_k$).
The orthogonal projection of $x_i$ onto the column space of $W_k$ is $\hat{x}_i = W_k W_k^\top x_i$. In matrix form:
$$\hat{X} = X W_k W_k^\top$$
The total squared reconstruction error is the squared Frobenius norm of the residual matrix:
$$\mathcal{E}(W_k) = \sum_{i=1}^n \|x_i - \hat{x}_i\|_2^2 = \|X - \hat{X}\|_F^2 = \operatorname{Tr}\Bigl( (X - X W_k W_k^\top)^\top (X - X W_k W_k^\top) \Bigr)$$
Expanding the product inside the trace:
$$\mathcal{E}(W_k) = \operatorname{Tr}\Bigl( X^\top X - X^\top X W_k W_k^\top - W_k W_k^\top X^\top X + W_k W_k^\top X^\top X W_k W_k^\top \Bigr)$$
Using the linearity of trace and its invariance under cyclic permutations ($\operatorname{Tr}(AB) = \operatorname{Tr}(BA)$):
1. $\operatorname{Tr}(W_k W_k^\top X^\top X) = \operatorname{Tr}(W_k^\top X^\top X W_k)$
2. $\operatorname{Tr}(X^\top X W_k W_k^\top) = \operatorname{Tr}(W_k^\top X^\top X W_k)$
3. $\operatorname{Tr}(W_k W_k^\top X^\top X W_k W_k^\top) = \operatorname{Tr}\Bigl( (W_k^\top X^\top X W_k) (W_k^\top W_k) \Bigr) = \operatorname{Tr}(W_k^\top X^\top X W_k)$ since $W_k^\top W_k = I_k$.

Substitute these back into the error expression:
$$\mathcal{E}(W_k) = \operatorname{Tr}(X^\top X) - 2\operatorname{Tr}(W_k^\top X^\top X W_k) + \operatorname{Tr}(W_k^\top X^\top X W_k) = \operatorname{Tr}(X^\top X) - \operatorname{Tr}(W_k^\top X^\top X W_k)$$
Notice that:
$$\operatorname{Tr}(X^\top X) = \sum_{j=1}^p \sigma_j^2$$
is the total variance (energy) of the dataset, which is fixed and completely independent of $W_k$.
Therefore, minimizing $\mathcal{E}(W_k)$ over all semi-orthogonal matrices $W_k$ is strictly equivalent to:
$$\max_{W_k \in \mathbb{R}^{p \times k}, \, W_k^\top W_k = I_k} \operatorname{Tr}(W_k^\top X^\top X W_k)$$
By the Poincaré Separation Theorem / Ky Fan Trace Maximization Theorem for symmetric matrices:
$$\operatorname{Tr}(W_k^\top X^\top X W_k) \le \sum_{j=1}^k \sigma_j^2$$
with equality holding if and only if the columns of $W_k$ span the eigenspace corresponding to the $k$ largest eigenvalues of $X^\top X$, which are the top-$k$ right singular vectors $V_k = [v_1, \dots, v_k]$.
Thus, choosing $W_k = V_k$:
$$\min_{W_k^\top W_k = I_k} \mathcal{E}(W_k) = \operatorname{Tr}(X^\top X) - \sum_{j=1}^k \sigma_j^2 = \sum_{j=1}^p \sigma_j^2 - \sum_{j=1}^k \sigma_j^2 = \sum_{j=k+1}^p \sigma_j^2$$
This matches the famous Eckart-Young-Mirsky Theorem, confirming that the truncated SVD projection $\hat{X}_k = X V_k V_k^\top = U_k \Sigma_k V_k^\top$ is the unique optimal rank-$k$ linear approximation of $X$. $\blacksquare$

---

### Sanity Checks & Edge Cases
1. **Full rank reconstruction ($k = p$):**
   When $k = p$, $V_p V_p^\top = I_p$.
   $$\mathcal{E}(V_p) = \sum_{j=p+1}^p \sigma_j^2 = 0 \implies \hat{X} = X$$
   Zero error, perfect reconstruction.
2. **Zero rank reconstruction ($k = 0$):**
   When $k = 0$, $\hat{X} = \mathbf{0}$.
   $$\mathcal{E}(\mathbf{0}) = \sum_{j=1}^p \sigma_j^2 = \|X\|_F^2 = \operatorname{Tr}(X^\top X)$$
   The reconstruction error equals the total squared variance of the dataset.
3. **Pythagorean decomposition of total variance:**
   Because the projection is orthogonal ($P_k = V_k V_k^\top$ satisfies $P_k^2 = P_k = P_k^\top$):
   $$\|X\|_F^2 = \|X P_k\|_F^2 + \|X (I - P_k)\|_F^2 = \sum_{j=1}^k \sigma_j^2 + \sum_{j=k+1}^p \sigma_j^2$$
   The explained variance plus the unexplained reconstruction error equals the total variance.

### Problem 6 [Python]

**Task:**
Implement PCA from scratch via the SVD and reproduce sklearn.decomposition.PCA on the digits dataset (agree to 10^-6 on the explained-variance ratios). Plot the cumulative explained variance and the 2-D projection coloured by class.


```python
# Problem 6 [Python]
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA

np.random.seed(42)

# Load digits dataset
digits = load_digits()
X, y = digits.data, digits.target
n_samples, n_features = X.shape

# 1. PCA from scratch via SVD
mean = np.mean(X, axis=0)
X_centered = X - mean
U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)

# Enforce deterministic coordinate signs matching scikit-learn convention
max_abs_cols = np.argmax(np.abs(Vt), axis=1)
signs = np.sign(Vt[range(Vt.shape[0]), max_abs_cols])
U *= signs
Vt *= signs[:, np.newaxis]

explained_variance = (S ** 2) / (n_samples - 1)
total_variance = np.sum(explained_variance)
explained_variance_ratio = explained_variance / total_variance
cumulative_variance = np.cumsum(explained_variance_ratio)
X_proj = X_centered @ Vt.T

# 2. Benchmark against sklearn.decomposition.PCA
sk_pca = PCA(n_components=n_features)
sk_X_proj = sk_pca.fit_transform(X)

# Verification: assert agreement to 10^-6 on explained variance ratios and variances
np.testing.assert_allclose(
    explained_variance_ratio,
    sk_pca.explained_variance_ratio_,
    rtol=1e-6,
    atol=1e-6,
    err_msg="Explained variance ratios between SVD scratch PCA and sklearn do not agree to 10^-6!"
)
np.testing.assert_allclose(
    explained_variance,
    sk_pca.explained_variance_,
    rtol=1e-6,
    atol=1e-6,
    err_msg="Sample explained variances do not agree to 10^-6!"
)
np.testing.assert_allclose(
    X_proj,
    sk_X_proj,
    rtol=1e-6,
    atol=1e-6,
    err_msg="Projected coordinates do not agree to 10^-6!"
)

max_ratio_diff = np.max(np.abs(explained_variance_ratio - sk_pca.explained_variance_ratio_))
print(f"[✓] PCA SVD scratch implementation verified against sklearn.decomposition.PCA")
print(f"    Maximum absolute difference in explained-variance ratios: {max_ratio_diff:.2e} (tolerance: 1.00e-06)")
print(f"    Variance explained by PC1: {explained_variance_ratio[0]*100:.2f}% | PC2: {explained_variance_ratio[1]*100:.2f}%")

# 3. Visualizations
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: Cumulative explained variance
axes[0].plot(range(1, n_features + 1), cumulative_variance, marker='o', markersize=3, color='#1f77b4', lw=2, label='Cumulative Explained Variance')
axes[0].axhline(y=0.90, color='#d62728', linestyle='--', alpha=0.8, label=f'90% Threshold (k = {np.searchsorted(cumulative_variance, 0.90) + 1})')
axes[0].axhline(y=0.95, color='#2ca02c', linestyle=':', alpha=0.8, label=f'95% Threshold (k = {np.searchsorted(cumulative_variance, 0.95) + 1})')
axes[0].set_xlabel('Number of Principal Components', fontsize=12)
axes[0].set_ylabel('Cumulative Explained Variance Ratio', fontsize=12)
axes[0].set_title('Cumulative Explained Variance vs. Dimensions', fontsize=13, fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend(loc='lower right', frameon=True)
axes[0].set_ylim([0, 1.05])

# Subplot 2: 2-D PCA Projection
scatter = axes[1].scatter(
    X_proj[:, 0],
    X_proj[:, 1],
    c=y,
    cmap=plt.get_cmap('tab10', 10),
    alpha=0.75,
    s=20,
    edgecolors='none'
)
axes[1].set_xlabel(f'Principal Component 1 ({explained_variance_ratio[0]*100:.1f}% var)', fontsize=12)
axes[1].set_ylabel(f'Principal Component 2 ({explained_variance_ratio[1]*100:.1f}% var)', fontsize=12)
axes[1].set_title('2-D PCA Projection of Digits Dataset (SVD Scratch)', fontsize=13, fontweight='bold')
cbar = plt.colorbar(scatter, ax=axes[1], ticks=range(10))
cbar.set_label('Digit Class', fontsize=11)
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

Run t-SNE and UMAP on the same features. Produce the three 2-D embeddings side by side (PCA, t-SNE, UMAP).


```python
# Problem 7 [Python]
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap

# Ensure reproducibility
np.random.seed(42)

# Load digits dataset
digits = load_digits()
X, y = digits.data, digits.target

# 1. PCA 2D embedding
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X)

# 2. t-SNE 2D embedding
tsne = TSNE(
    n_components=2,
    perplexity=30.0,
    random_state=42,
    init='pca',
    learning_rate='auto'
)
X_tsne = tsne.fit_transform(X)

# 3. UMAP 2D embedding
um = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    random_state=42
)
X_umap = um.fit_transform(X)

# Plot the three embeddings side by side
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))
cmap = plt.get_cmap('tab10', 10)

embeddings = [
    (X_pca, 'PCA (Linear Projection)', 'Principal Component 1', 'Principal Component 2'),
    (X_tsne, 't-SNE (Perplexity=30)', 't-SNE Dimension 1', 't-SNE Dimension 2'),
    (X_umap, 'UMAP (n_neighbors=15, min_dist=0.1)', 'UMAP Dimension 1', 'UMAP Dimension 2')
]

for idx, (emb, title, xlabel, ylabel) in enumerate(embeddings):
    scatter = axes[idx].scatter(
        emb[:, 0],
        emb[:, 1],
        c=y,
        cmap=cmap,
        alpha=0.75,
        s=16,
        edgecolors='none'
    )
    axes[idx].set_title(title, fontsize=13, fontweight='bold')
    axes[idx].set_xlabel(xlabel, fontsize=11)
    axes[idx].set_ylabel(ylabel, fontsize=11)
    axes[idx].grid(True, linestyle='--', alpha=0.4)

# Shared colorbar for digit classes
cbar = fig.colorbar(scatter, ax=axes, orientation='horizontal', fraction=0.04, pad=0.15, ticks=range(10))
cbar.set_label('Digit Class (0–9)', fontsize=12)
cbar.set_ticklabels([f'Digit {i}' for i in range(10)])

plt.suptitle('Comparison of 2-D Dimensionality Reduction Techniques on Digits Dataset', fontsize=15, fontweight='bold', y=1.02)
plt.show()
```

### Problem 8 [Analysis]

**Task:**
Compare the three embeddings from Problem 7: which one preserves global distances, which preserves local neighbourhoods, and why can t-SNE/UMAP distances not be read as similarities? Relate your answer to Problems 1–3.

---

### Empirical Observations from Problem 7
The three 2-D embeddings of the 64-dimensional handwritten digits dataset display strikingly different topological properties:
1. **PCA (Linear Projection):** Produces a single continuous cloud with heavy class overlap. While certain visually distinct digits (such as 0, 4, and 6) exhibit partial separation along the periphery, the majority of digits (1, 2, 3, 5, 7, 8, and 9) form a dense, entangled continuum in the centre.
2. **t-SNE (Probabilistic Manifold Embedding):** Completely resolves the overlap, decomposing the dataset into ten well-isolated, compact clusters with wide white-space margins separating the digit classes.
3. **UMAP (Fuzzy Simplicial Set Embedding):** Yields even sharper, tighter clusters than t-SNE. Moreover, it places digits with related stroke topologies (e.g., 1 and 7, or 3, 5, and 8) into closer proximity, suggesting an improved preservation of intermediate global continuum.

---

### Global Distances vs. Local Neighbourhood Preservation

#### 1. Global Distance Preservation: PCA
- **Mechanism:** PCA performs a linear, orthogonal projection $X_{\mathrm{proj}} = X V_2$ onto the 2-D subspace spanned by the top two eigenvectors of the sample covariance matrix. As proven in **Problem 4**, this projection globally maximizes sample variance and minimizes the total sum of squared Euclidean reconstruction errors across all data points.
- **Consequence:** PCA faithfully preserves large-scale, macroscopic Euclidean distances. If two clusters are distant along the primary directions of variance, they remain distant in the PCA projection.
- **Limitation:** As revealed by the cumulative explained variance curve in **Problem 6**, the first two principal components account for only $\approx 28.5\%$ of the total data variance ($14.9\%$ from PC1 and $13.6\%$ from PC2); 29 components are required to capture $95\%$ of the variance. Consequently, squashing 64 dimensions into 2 discards over $71\%$ of the variance, forcing orthogonally separated manifolds to collapse on top of each other.

#### 2. Local Neighbourhood Preservation: t-SNE and UMAP
- **t-SNE:** Constructs high-dimensional pairwise probabilities $p_{j|i}$ using an adaptive Gaussian kernel, and low-dimensional probabilities $q_{ij}$ using a heavy-tailed Cauchy (Student-$t$ with 1 degree of freedom) distribution. It minimizes the Kullback-Leibler divergence:
  $$\mathrm{KL}(P \parallel Q) = \sum_{i \neq j} p_{ij} \log \frac{p_{ij}}{q_{ij}}$$
  The asymmetry of KL divergence heavily penalizes **false negatives** ($p_{ij}$ large but $q_{ij}$ small—i.e., true high-dimensional neighbours placed far apart in 2D), forcing local neighbourhoods to remain intact. However, if two points are distant in high dimensions ($p_{ij} \approx 0$), the penalty is negligible regardless of where they land in low-dimensional space. Thus, t-SNE aggressively preserves local structure at the expense of global geometry.
- **UMAP:** Models the underlying manifold using fuzzy simplicial sets and minimizes fuzzy set cross-entropy:
  $$C(P, Q) = \sum_{i \neq j} \left[ p_{ij} \log \frac{p_{ij}}{q_{ij}} + (1 - p_{ij}) \log \frac{1 - p_{ij}}{1 - q_{ij}} \right]$$
  The second term $((1 - p_{ij}) \log \frac{1}{1 - q_{ij}})$ exerts a continuous repulsive force between non-neighbouring points, which preserves more macroscopic topological relationships and cluster continuum than t-SNE while maintaining comparable local clustering.

### Why t-SNE and UMAP Distances Cannot Be Read as Similarities

Interpreting Euclidean distances in a 2-D t-SNE or UMAP plot as direct indicators of high-dimensional similarity is fundamentally flawed due to three mathematical factors:

1. **Adaptive Local Bandwidths ($\sigma_i$ and $\rho_i$):**
   Both algorithms locally re-scale distances. In t-SNE, the Gaussian bandwidth $\sigma_i$ is chosen dynamically for each point $x_i$ to satisfy a fixed perplexity ($2^{H(P_i)} = \text{Perp}$); in UMAP, $\rho_i$ represents the distance to the nearest neighbour. Consequently, **dense clusters are artificially expanded** (large internal low-dimensional spacing) while **sparse clusters are compressed**. A 2-cm separation inside a dense cluster may correspond to a much smaller high-dimensional distance than a 1-cm separation in a sparse region.
2. **Crowding Problem and Heavy-Tailed Distortions:**
   In high dimensions, volume grows exponentially with radius (as shown in Problem 3), allowing many mutually equidistant neighbours. In 2 dimensions, space is severely restricted. Both algorithms resolve this "crowding problem" by non-linearly distorting distances via heavy-tailed kernels ($q_{ij} \propto (1 + \|y_i - y_j\|^2)^{-1}$). This non-linear mapping destroys linear proportionality: doubling the distance in the embedding does not imply that the samples are twice as dissimilar in feature space.
3. **Inter-Cluster Distances Are Optimization Artifacts:**
   Because long-range attraction is virtually zero, the relative distances between separate clusters are driven by random initialization, gradient step sizes, and the repulsion phase of the optimization. Two clusters situated on opposite sides of the plot are not necessarily more dissimilar than two adjacent clusters; only within-cluster connectivity is topologically meaningful.

### Synthesis and Connections to Problems 1–3

#### Connection to Problem 1 (Euclidean vs. Cosine Geometry on Unit Spheres)
- In **Problem 1**, we proved that on the unit sphere ($\|x\|_2 = \|y\|_2 = 1$), Euclidean distance and cosine similarity induce the exact same neighbour ranking ($\|x - y\|_2^2 = 2(1 - \cos \theta)$).
- High-dimensional image vectors are often subject to uninformative variations in overall pixel brightness or stroke thickness (vector magnitude). When data is normalized to unit norm, the nearest-neighbour graph constructed by t-SNE and UMAP becomes equivalent under both Euclidean and cosine metrics. This confirms that neighbourhood-preserving algorithms rely fundamentally on **angular and directional geometry** rather than ambient Euclidean scale.

#### Connection to Problem 2 (Non-Metric Distances and Triangle Inequality Violations)
- In **Problem 2**, we proved that the cosine distance $d_{\cos} = 1 - \cos \theta$ is not a valid metric because it violates the triangle inequality due to its strict convexity ($d_{\cos} = 2\sin^2(\theta/2)$).
- Similarly, the effective distance in the low-dimensional embedding spaces of t-SNE and UMAP is non-metric relative to the original space. The non-linear warping and manifold tearing mean that if $d(A, B)$ is small and $d(B, C)$ is small in high dimensions, $A$ and $C$ can be torn apart into different clusters, violating metric transitivity and the triangle inequality across cluster boundaries.

#### Connection to Problem 3 (Curse of Dimensionality and Distance Concentration)
- In **Problem 3**, we demonstrated that as ambient dimension $d \to \infty$, the relative distance gap $(D_{\max} - D_{\min})/D_{\min} \xrightarrow{p} 0$ and the volume of a ball concentrates in an infinitesimal boundary shell ($1 - (1 - \varepsilon)^d \to 1$).
- In 64-dimensional space, raw Euclidean distances suffer from this concentration: pairwise distances between digits crowd into a narrow numerical band, severely diminishing distance contrast. This explains why linear projections like PCA struggle to isolate classes: the background distance noise drowns out the signal.
- t-SNE and UMAP directly overcome this curse by converting absolute high-dimensional Euclidean distances into **relative local probabilities** with point-specific bandwidths $\sigma_i$. By normalizing locally, they ignore the uninformative global distance concentration and extract only the local rank-order relationships that define the true underlying handwritten manifold.
